```python
def preprocess_text(text):
    if isinstance(text, str):
        # Remove URLs
        text = re.sub(r'https?://\S+', '', text)
        # Remove HTML tags
        text = re.sub(r'<.*?>', '', text)
        # Remove extra whitespaces
        text = re.sub(r'\s+', ' ', text).strip()
    else:
        text = ''
    return text
```

```python
from indicnlp.tokenize import indic_tokenize
import string
```

```python
def tokenize_data(text):
  tokens = indic_tokenize.trivial_tokenize(text, lang='mr')
  marathi_punct = '।ः”“‘’–…'
  tokens = [t for t in tokens if t not in string.punctuation and t not in marathi_punct]
  marathi_stopwords = {
      'आहे', 'असे', 'आणि', 'किंवा', 'तर', 'वर', 'च्या', 'चे', 'ची', 'मध्ये', 'होते', 'होणार',
      'तो', 'ही', 'ते', 'या', 'तो', 'त्या', 'म्हणून', 'नाही', 'झाले', 'झालेल्या', 'एक'
  }
  tokens = [t for t in tokens if t not in marathi_stopwords]
  suffixes = ['ला', 'नी', 'च्या', 'चे', 'ची', 'मध्ये', 'वर', 'तील', 'ता', 'ते', 'तो', 'ली', 'ला']
  def simple_lemmatize(word):
      for suf in suffixes:
          if word.endswith(suf):
              return word[:-len(suf)]
      return word

  lemmatized_tokens = [simple_lemmatize(t) for t in tokens]

  return lemmatized_tokens
```

```python
df['tokens'] = df['clean_content'].apply(tokenize_data)
```

```python
from itertools import chain

# Flatten list of token lists into one big list
all_tokens = list(chain.from_iterable(df['tokens']))
```

```python
import nltk
from nltk import ngrams
from collections import Counter

# Function to get top n-grams
def get_top_ngrams(tokens, n=2, top_k=20):
    n_grams = ngrams(tokens, n)
    return Counter(n_grams).most_common(top_k)

# Example: Top 20 bigrams
top_bigrams = get_top_ngrams(all_tokens, n=2, top_k=20)
print("Top Bigrams:", top_bigrams)

# Example: Top 20 trigrams
top_trigrams = get_top_ngrams(all_tokens, n=3, top_k=20)
print("Top Trigrams:", top_trigrams)

```

```python
from collections import defaultdict
from nltk import trigrams
tri_grams = list(trigrams(all_tokens))

model = defaultdict(lambda: defaultdict(lambda: 0))
for w1, w2, w3 in tri_grams:
    model[(w1, w2)][w3] += 1

for w1_w2 in model:
    total_count = float(sum(model[w1_w2].values()))
    for w3 in model[w1_w2]:
        model[w1_w2][w3] /= total_count


def predict_next_word(w1, w2):
    next_word_probs = model[w1, w2]
    if next_word_probs:
        return max(next_word_probs, key=next_word_probs.get)
    else:
        return "No prediction available"
```

```python
class Node:
    def __init__(self, symbol, children=None, word=None):
        self.symbol = symbol
        self.children = children or []
        self.word = word  # for terminals

    def __repr__(self, level=0):
        indent = "  " * level
        if self.word:
            return f"{indent}{self.symbol} -> '{self.word}'\n"
        else:
            result = f"{indent}{self.symbol}\n"
            for child in self.children:
                result += child.__repr__(level + 1)
            return result
```

```python
class TopDownParser:
    def __init__(self, grammar, start_symbol):
        self.grammar = grammar  # dict: symbol -> list of productions (each a list of symbols)
        self.start_symbol = start_symbol

    def parse(self, tokens):
        self.tokens = tokens
        self.pos = 0
        node = self._parse_symbol(self.start_symbol)
        if node and self.pos == len(tokens):
            return node
        else:
            raise Exception("Parsing failed or incomplete parse")

    def _parse_symbol(self, symbol):
        # Terminal symbol
        if symbol not in self.grammar:
            if self.pos < len(self.tokens) and self.tokens[self.pos] == symbol:
                node = Node(symbol, word=self.tokens[self.pos])
                self.pos += 1
                return node
            else:
                return None

        # Non-terminal: try all productions
        for production in self.grammar[symbol]:
            start_pos = self.pos
            children = []
            for sym in production:
                child = self._parse_symbol(sym)
                if child is None:
                    self.pos = start_pos
                    children = None
                    break
                children.append(child)
            if children is not None:
                return Node(symbol, children)
        return None


if __name__ == "__main__":
    grammar = {
        'S': [['NP', 'VP']],
        'NP': [['Det', 'N']],
        'VP': [['V', 'NP']],
        'Det': [['the'], ['a']],
        'N': [['cat'], ['dog']],
        'V': [['chases'], ['sees']]
    }

    sentence = "the cat chases a dog"
    tokens = sentence.split()

    parser = TopDownParser(grammar, 'S')
    parse_tree = parser.parse(tokens)
    print(parse_tree)
```
